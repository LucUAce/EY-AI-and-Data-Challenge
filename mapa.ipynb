```
import folium
import pandas as pd

# 1. Leer el fichero de coordenadas, tmb con water_quality_training_dataset.csv
df = pd.read_csv('submission_template.csv')

df = df[['Longitude', 'Latitude', 'Sample Date']]

display(df.head())
# 2. Crear el mapa centrado en una ubicación
mapa = folium.Map(location=[40.4167, -3.7033], zoom_start=6)

# 3. Añadir puntos al mapa
for index, row in df.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=row['Sample Date'],
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(mapa)
```

```
mapa.save("/tmp/mapa_puntos_sub.html")

import snowflake
from snowflake.snowpark.context import get_active_session
session = get_active_session()

session.sql(f"""
    PUT file:///tmp/mapa_puntos_sub.html
    snow://workspace/USER$.PUBLIC.DEFAULT$/versions/live/
    AUTO_COMPRESS=FALSE
    OVERWRITE=TRUE
""").collect()
```
